```python
import pandas as pd
df=pd.read_csv('../data/Cleaned/Superstore sales cleaned.csv')
```

```python
total_sales=df['Sales'].sum()
total_profit= df['Profit'].sum()
overall_margin = (total_profit/total_sales)*100

print("Total Sales:",total_sales)
print("Total Profit:",total_profit)
print("overall margin:",overall_margin)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#PROFIT BY CATEGORY
category_profit =df.groupby('Category')['Profit'].sum()
print(category_profit)
```

```python
category_profit.sort_values().plot(kind='barh')
plt.title('Total profit by category')
plt.xlabel('Profit($)')
plt.show()
```

```python
#profit by sub category
subcat_profit=df.groupby('Sub-Category')['Profit'].sum().sort_values()
print(subcat_profit)
```

```python
subcat_profit.plot(kind='barh')
plt.title('Total profit by sub category')
plt.xlabel('Profit($)')
plt.show()
```

```python
discount_profit=df.groupby('Discount')['Profit'].mean().sort_index()
print(discount_profit)
```

```python
#profit by region
region_profit=df.groupby('Region')['Profit'].sum().sort_values()
print(region_profit)
```

```python
region_profit.plot(kind='barh')
plt.title('total Profit by region')
plt.xlabel('Profit')
plt.show()
```
